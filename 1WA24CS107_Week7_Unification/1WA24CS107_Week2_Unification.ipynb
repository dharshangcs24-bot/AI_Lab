```python

# Unification in First Order Logic

def is_variable(x):
    return isinstance(x, str) and x in ['x', 'y', 'z', 'X', 'Y', 'Z']

def apply_substitution(term, subs):
    if is_variable(term):
        while term in subs:
            term = subs[term]
        return term

    if isinstance(term, tuple):
        return tuple(apply_substitution(t, subs) for t in term)

    return term

def occurs_check(var, term, subs):
    term = apply_substitution(term, subs)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, t, subs) for t in term)

    return False

def unify(x, y, subs):
    x = apply_substitution(x, subs)
    y = apply_substitution(y, subs)

    if x == y:
        return True

    if is_variable(x):
        if occurs_check(x, y, subs):
            return False
        subs[x] = y
        return True

    if is_variable(y):
        if occurs_check(y, x, subs):
            return False
        subs[y] = x
        return True

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y) or x[0] != y[0]:
            return False

        for a, b in zip(x[1:], y[1:]):
            if not unify(a, b, subs):
                return False

        return True

    return False


# Main Program

print("UNIFICATION IN FIRST ORDER LOGIC")

expr1 = ("Eats", "x", "Apple")
expr2 = ("Eats", "Riya", "y")

print("\nFirst Expression:", expr1)
print("Second Expression:", expr2)

subs = {}

if unify(expr1, expr2, subs):
    print("\nUnification Successful!")
    print("\nMost General Unifier (MGU):")

    for var in subs:
        print(var, "=", apply_substitution(subs[var], subs))

    print("\nUnified Expression:")
    print(apply_substitution(expr1, subs))
    print(apply_substitution(expr2, subs))

else:
    print("\nUnification Failed!")
```
